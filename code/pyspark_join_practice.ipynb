```python
dataj1 = [('1','gaur','d01'),
          ('2','kit','d02'),
          ('3','sam','d03'),
          ('4','tim','d03'),
          ('5','aman','d05'),
          ('6','nad','d06')] 

schemaj1 = 'emp_id STRING, emp_name STRING, dept_id STRING' 

df1 = spark.createDataFrame(dataj1,schemaj1)

dataj2 = [('d01','HR'),
          ('d02','Marketing'),
          ('d03','Accounts'),
          ('d04','IT'),
          ('d05','Finance')]

schemaj2 = 'dept_id STRING, department STRING'

df2 = spark.createDataFrame(dataj2,schemaj2)
```

```python
df1.display()
```

```python
df2.display()
```

### Inner join

```python
ijoined_df = df1.join(df2, df1['dept_id'] == df2['dept_id'], 'inner').drop(df2.dept_id)
ijoined_df.display()
```

### Left Join

```python
ljoined_df = df1.join(df2, df1['dept_id'] == df2['dept_id'], 'left')
ljoined_df.display()
```

### Right Join

```python
rjoined_df = df1.join(df2,df1['dept_id'] == df2['dept_id'], 'right')
rjoined_df.display()
```

### Anti Join

```python
ajoined_df = df1.join(df2, df1['dept_id'] == df2['dept_id'], 'anti')
ajoined_df.display()
```

```python
ajoined_df = df1.join(df2, df1['dept_id'] == df2['dept_id'], 'left').filter(df2.dept_id.isNull())
ajoined_df.display()
```

##Windows Function

###Row_number

```python
df = spark.read.option('header', 'true').option('inferSchema', 'true').csv('/Workspace/Users/pankaj_databricks_outlook.com#ext#@pankajdatabricksoutlook.onmicrosoft.com/PySpark_Practice/data/BigMart Sales.csv')
```

```python
df.limit(5).display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
```

```python


df.withColumn('rowNum', row_number().over(Window.orderBy('Item_Identifier'))).display()
```

###Rank

```python
df.withColumn('rank', rank().over(Window.orderBy('Item_Identifier'))).display()
```

```python
df.withColumn('rank', rank().over(Window.orderBy(col('Item_Identifier').desc()))).display()
```

###Dense Rank

```python
df.withColumn('rank', rank().over(Window.orderBy('Item_Identifier')))\
    .withColumn('dense_rank',dense_rank().over(Window.orderBy('Item_identifier'))).display()
```

## Cumulative Sum

```python
sum_window = Window.orderBy('Item_Type').rowsBetween(Window.unboundedPreceding, Window.currentRow)
df.withColumn('cumsum', sum('Item_MRP').over(sum_window)).display()
```
