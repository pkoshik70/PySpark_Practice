**StructType() Schema**

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
my_struct_schema = StructType([
    StructField('Item_Identifier', StringType(), True),
    StructField('Item_Weight', StringType(), True),
    StructField('Item_Fat_Content', StringType(), True),
    StructField('Item_Visibility', StringType(), True),
    StructField('Item_Type', StringType(), True),
    StructField('Item_MRP', StringType(), True),
    StructField('Outlet_Identifier', StringType(), True),
    StructField('Outlet_Establishment_Year', StringType(), True),
    StructField('Outlet_Size', StringType(), True),
    StructField('Outlet_Location_Type', StringType(), True),
    StructField('Outlet_Type', StringType(), True),
    StructField('Item_Outlet_Sales', StringType(), True)
])
```

###Union

```python
data1 = [('1','kad'),
        ('2','sid')]
schema1 = 'id STRING, name STRING' 

df1 = spark.createDataFrame(data1,schema1)

data2 = [('3','rahul'),
        ('4','jas')]
schema2 = 'id STRING, name STRING' 

df2 = spark.createDataFrame(data2,schema2)
```

```python
df1 = df1.withColumn('cur_date',current_date())
df1.display()
```

```python
df2 = df1.withColumn('cur_date',date_add('cur_date',7))
df2.display()
```

```python
df2 = df1.withColumn('cur_date',date_sub('cur_date',7))
df2.display()
```

```python
df2 = df1.withColumn('cur_date', date_add('cur_date',-7))
df2.display()
```

##DateDiff

```python
df1 = df1.withColumn('week_after',date_add('cur_date',7)).withColumn('week_before', date_add('cur_date',-7))
df1.display()
```

```python
df1 = df1.withColumn('datediff', datediff('cur_date', 'week_after'))
df1.display()
```
