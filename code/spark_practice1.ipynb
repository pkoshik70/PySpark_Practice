## Split And Indexing

```python
from pyspark.sql.functions import *
```

```python
df2 = df.withColumn('Outlet_Type', split('Outlet_Type', ' '))
```

```python
df1 = df.withColumn('Outlet_Type_Category', split(col('Outlet_Type'), ' ')[1]).withColumn('Outlet_Type', split(col('Outlet_Type'), ' ')[0])
df1.limit(5).display()
```

```python
df2.withColumn('Outlet_Type', explode('Outlet_type')).limit(5).display()
```

```python
df.limit(5).display()
```

```python
df2.withColumn('Type1_flag', array_contains('Outlet_Type', 'Type1')).limit(5).display()
```

```python
df.groupBy(col('Item_Type')).agg(sum('Item_MRP')).limit(10).display()
```

```python
df.groupBy(col('Item_Type')).agg(avg('Item_MRP')).limit(10).display()
```

```python
df.groupBy(col('Item_Type'),col('Outlet_Size')).agg(sum('Item_MRP')).filter(col('Outlet_Size').isNotNull()).orderBy('Item_Type','Outlet_Size').display()
```

```python
df.groupBy('Item_Type', 'Outlet_Size').agg(sum('Item_MRP'),avg('Item_MRP')).filter(col('Outlet_Size').isNotNull()).display()
```
